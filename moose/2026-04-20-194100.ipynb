```python
import netCDF4 as nc
import numpy as np

fn = "coupled_single_T.e"
ds = nc.Dataset(fn)

x = np.array(ds.variables["coordx"][:], dtype=float)
y = np.array(ds.variables["coordy"][:], dtype=float)
T = np.array(ds.variables["vals_nod_var1"][-1], dtype=float)

ks = 0.50
kf = 0.15
dx = 1.0 / 80.0
x_if = 0.4

ys = np.linspace(0.05, 0.95, 20)

for yy in ys:
    # nearest nodes at left, interface, right
    i_left = np.argmin((x - (x_if - dx))**2 + (y - yy)**2)
    i_if   = np.argmin((x - x_if)**2 + (y - yy)**2)
    i_right= np.argmin((x - (x_if + dx))**2 + (y - yy)**2)

    T_left = T[i_left]
    T_if = T[i_if]
    T_right = T[i_right]

    # Approximate normal derivative from each side
    grad_left = (T_if - T_left) / dx
    grad_right = (T_right - T_if) / dx

    flux_left = ks * grad_left
    flux_right = kf * grad_right

    print(
        f"y={yy:.3f}, "
        f"T_left={T_left:.6f}, T_if={T_if:.6f}, T_right={T_right:.6f}, "
        f"ks*dTdx_left={flux_left:.6f}, "
        f"kf*dTdx_right={flux_right:.6f}"
    )
```

```python
import netCDF4 as nc
import numpy as np

fn = "coupled_single_T.e"
ds = nc.Dataset(fn)

x = np.array(ds.variables["coordx"][:], dtype=float)
y = np.array(ds.variables["coordy"][:], dtype=float)
time = float(ds.variables["time_whole"][-1])

# variable order from your output:
# vals_nod_var1 = T
# vals_nod_var2 = phi
phi = np.array(ds.variables["vals_nod_var2"][-1], dtype=float)

# Helper to decode Exodus names
def decode_names(arr):
    names = []
    for row in arr[:]:
        chars = []
        for c in row:
            if isinstance(c, bytes):
                if c != b'\x00' and c != b'':
                    chars.append(c.decode("utf-8", errors="ignore"))
            else:
                # masked values may appear
                try:
                    if c is not np.ma.masked:
                        chars.append(str(c))
                except Exception:
                    pass
        names.append("".join(chars).strip())
    return names

ns_names = decode_names(ds.variables["ns_names"])
print("node set names:")
for i, name in enumerate(ns_names, start=1):
    print(i, name)

# find fuel_left node set
idx = None
for i, name in enumerate(ns_names, start=1):
    if name == "fuel_left":
        idx = i
        break

if idx is None:
    raise RuntimeError("fuel_left node set not found")

# Exodus node sets are 1-based
node_ids = np.array(ds.variables[f"node_ns{idx}"][:], dtype=int) - 1

yy = y[node_ids]
phi_left = phi[node_ids]

# exact final-time BC
bc = (1.0 - np.exp(-5.0 * time)) * (1.0 + 0.2 * np.sin(np.pi * yy)**2)

diff = np.abs(phi_left - bc)

print("time:", time)
print("fuel_left node count:", len(node_ids))
print("max |phi_left - BC| on exact nodes:", diff.max())
print("mean |phi_left - BC| on exact nodes:", diff.mean())

# print a few rows
order = np.argsort(yy)
for j in order[::max(1, len(order)//8)]:
    print(f"y={yy[j]:.6f}, phi={phi_left[j]:.8f}, BC={bc[j]:.8f}, diff={diff[j]:.3e}")
```

```python
import netCDF4 as nc
import numpy as np
from scipy.interpolate import LinearNDInterpolator

fn = "coupled_single_T_adv_80x64.e"
ds = nc.Dataset(fn)

# decode Exodus nodal variable names
raw = ds.variables["name_nod_var"][:]

def decode_names(raw):
    names = []
    for row in raw:
        chars = []
        for c in row:
            if c is np.ma.masked:
                continue
            if isinstance(c, bytes):
                chars.append(c.decode("utf-8"))
            else:
                chars.append(str(c))
        names.append("".join(chars).strip("\x00").strip())
    return names

names = decode_names(raw)
print("decoded nodal vars:", names)

# coords
x = np.array(ds.variables["coordx"][:], dtype=float)
y = np.array(ds.variables["coordy"][:], dtype=float)
t = np.array(ds.variables["time_whole"][:], dtype=float)

# map variable names -> vals_nod_var*
name_to_idx = {name: i + 1 for i, name in enumerate(names)}

T_all = np.array(ds.variables[f"vals_nod_var{name_to_idx['T']}"][:], dtype=float)
phi_all = np.array(ds.variables[f"vals_nod_var{name_to_idx['phi']}"][:], dtype=float)

T = T_all[-1]
phi = phi_all[-1]
t_final = float(t[-1])

print("final T min/max:", T.min(), T.max())
print("final phi min/max:", phi.min(), phi.max())

pts = np.column_stack([x, y])
interp_T = LinearNDInterpolator(pts, T)
interp_phi = LinearNDInterpolator(pts, phi)

x_unique = np.unique(np.round(x, 10))
y_unique = np.unique(np.round(y, 10))
dx = np.min(np.diff(x_unique))
dy = np.min(np.diff(y_unique))
eps_x = dx / 2.0
eps_y = dy / 2.0

print("dx =", dx, "dy =", dy)

# neutron BC check on left
ys = np.linspace(0.05, 0.95, 100)
phi_left = np.array([interp_phi(0.0, yy) for yy in ys], dtype=float)
phi_bc = np.array([
    (1 - np.exp(-5*t_final)) * (1 + 0.2*np.sin(np.pi*yy)**2)
    for yy in ys
])
print("max |phi(left)-BC|:", np.nanmax(np.abs(phi_left - phi_bc)))
print("mean |phi(left)-BC|:", np.nanmean(np.abs(phi_left - phi_bc)))

# neutron top/bottom zero-flux check
xs = np.linspace(0.05, 0.35, 50)
phi_top = np.array([interp_phi(xx, 1.0) for xx in xs], dtype=float)
phi_top_in = np.array([interp_phi(xx, 1.0 - eps_y) for xx in xs], dtype=float)
dphidy_top = (phi_top - phi_top_in) / eps_y

phi_bot = np.array([interp_phi(xx, 0.0) for xx in xs], dtype=float)
phi_bot_in = np.array([interp_phi(xx, eps_y) for xx in xs], dtype=float)
dphidy_bot = (phi_bot_in - phi_bot) / eps_y

print("max |dphi/dy| top interior:", np.nanmax(np.abs(dphidy_top)))
print("mean |dphi/dy| top interior:", np.nanmean(np.abs(dphidy_top)))
print("max |dphi/dy| bottom interior:", np.nanmax(np.abs(dphidy_bot)))
print("mean |dphi/dy| bottom interior:", np.nanmean(np.abs(dphidy_bot)))

# temperature near interface
x_if = 0.4
ys = np.linspace(0.05, 0.95, 100)

T_left = np.array([interp_T(x_if - eps_x, yy) for yy in ys], dtype=float)
T_if = np.array([interp_T(x_if, yy) for yy in ys], dtype=float)
T_right = np.array([interp_T(x_if + eps_x, yy) for yy in ys], dtype=float)

print("max |T_left - T_right| near interface:", np.nanmax(np.abs(T_left - T_right)))
print("mean |T_left - T_right| near interface:", np.nanmean(np.abs(T_left - T_right)))

ks = 0.5
kf = 0.15
flux_left = ks * (T_if - T_left) / eps_x
flux_right = kf * (T_right - T_if) / eps_x

print("max |flux_left - flux_right|:", np.nanmax(np.abs(flux_left - flux_right)))
print("mean |flux_left - flux_right|:", np.nanmean(np.abs(flux_left - flux_right)))

# advection check: coolant should warm as y increases
xs_cool = np.linspace(0.5, 0.8, 25)  # stay away from interface and right wall a bit
y_low = 0.2
y_mid = 0.5
y_high = 0.8

T_low = np.array([interp_T(xx, y_low) for xx in xs_cool], dtype=float)
T_mid = np.array([interp_T(xx, y_mid) for xx in xs_cool], dtype=float)
T_high = np.array([interp_T(xx, y_high) for xx in xs_cool], dtype=float)

print("coolant mean T at y=0.2:", np.nanmean(T_low))
print("coolant mean T at y=0.5:", np.nanmean(T_mid))
print("coolant mean T at y=0.8:", np.nanmean(T_high))
```

```python
import netCDF4 as nc
import numpy as np
from scipy.interpolate import LinearNDInterpolator

fn = "coupled_single_T_adv_80x64.e"   # change if your file name differs
ds = nc.Dataset(fn)

x = np.array(ds.variables["coordx"][:], dtype=float)
y = np.array(ds.variables["coordy"][:], dtype=float)
t = np.array(ds.variables["time_whole"][:], dtype=float)

# Your file order:
# vals_nod_var1 = T
# vals_nod_var2 = phi
T_all   = np.array(ds.variables["vals_nod_var1"][:], dtype=float)
phi_all = np.array(ds.variables["vals_nod_var2"][:], dtype=float)

T = T_all[-1]
phi = phi_all[-1]
t_final = t[-1]

pts = np.column_stack([x, y])
interp_T = LinearNDInterpolator(pts, T)
interp_phi = LinearNDInterpolator(pts, phi)

x_unique = np.unique(np.round(x, 10))
y_unique = np.unique(np.round(y, 10))

dx = np.min(np.diff(x_unique))
dy = np.min(np.diff(y_unique))

print("dx, dy:", dx, dy)
print("final phi min/max:", phi.min(), phi.max())
print("final T min/max:", T.min(), T.max())

# -------------------------
# 1. phi left BC check
# -------------------------
ys = np.linspace(0.05, 0.95, 100)

phi_left = np.array([interp_phi(0.0, yy) for yy in ys])
phi_bc = np.array([
    (1 - np.exp(-5*t_final)) * (1 + 0.2*np.sin(np.pi*yy)**2)
    for yy in ys
])

print("max |phi(left)-BC|:", np.nanmax(np.abs(phi_left - phi_bc)))
print("mean |phi(left)-BC|:", np.nanmean(np.abs(phi_left - phi_bc)))

# -------------------------
# 2. phi top/bottom zero flux
# -------------------------
eps_y = dy / 2
xs_fuel = np.linspace(0.05, 0.35, 100)

phi_top = np.array([interp_phi(xx, 1.0) for xx in xs_fuel])
phi_top_in = np.array([interp_phi(xx, 1.0 - eps_y) for xx in xs_fuel])
dphidy_top = (phi_top - phi_top_in) / eps_y

phi_bot = np.array([interp_phi(xx, 0.0) for xx in xs_fuel])
phi_bot_in = np.array([interp_phi(xx, eps_y) for xx in xs_fuel])
dphidy_bot = (phi_bot_in - phi_bot) / eps_y

print("max |dphi/dy| top:", np.nanmax(np.abs(dphidy_top)))
print("mean |dphi/dy| top:", np.nanmean(np.abs(dphidy_top)))
print("max |dphi/dy| bottom:", np.nanmax(np.abs(dphidy_bot)))
print("mean |dphi/dy| bottom:", np.nanmean(np.abs(dphidy_bot)))

# -------------------------
# 3. phi interface zero flux
# -------------------------
eps_x = dx / 2
x_if = 0.4

ys_if = np.linspace(0.05, 0.95, 100)
phi_if = np.array([interp_phi(x_if, yy) for yy in ys_if])
phi_if_left = np.array([interp_phi(x_if - eps_x, yy) for yy in ys_if])
dphidx_if = (phi_if - phi_if_left) / eps_x

D_phi = 0.05
print("max |dphi/dx| interface:", np.nanmax(np.abs(dphidx_if)))
print("max neutron flux |D*dphi/dx| interface:", D_phi*np.nanmax(np.abs(dphidx_if)))

# -------------------------
# 4. T interface smoothness
# -------------------------
T_left = np.array([interp_T(x_if - eps_x, yy) for yy in ys_if])
T_if   = np.array([interp_T(x_if, yy) for yy in ys_if])
T_right = np.array([interp_T(x_if + eps_x, yy) for yy in ys_if])

print("max |T_left-T_right| near interface:", np.nanmax(np.abs(T_left - T_right)))
print("mean |T_left-T_right| near interface:", np.nanmean(np.abs(T_left - T_right)))

# -------------------------
# 5. advection direction check
# -------------------------
xs_cool = np.linspace(0.45, 0.95, 100)

for yy in [0.2, 0.5, 0.8]:
    vals = np.array([interp_T(xx, yy) for xx in xs_cool])
    print(f"coolant mean T at y={yy}:", np.nanmean(vals))

# -------------------------
# 6. inlet check: bottom coolant should be near zero
# -------------------------
T_bottom_cool = np.array([interp_T(xx, 0.0) for xx in xs_cool])
print("max |T| at coolant bottom inlet:", np.nanmax(np.abs(T_bottom_cool)))
print("mean |T| at coolant bottom inlet:", np.nanmean(np.abs(T_bottom_cool)))
```

```python
import netCDF4 as nc
import numpy as np

fn = "coupled_single_T_adv_80x64.e"   # your file
ds = nc.Dataset(fn)

# coordinates
x = np.array(ds.variables["coordx"][:], dtype=float)
y = np.array(ds.variables["coordy"][:], dtype=float)
t = np.array(ds.variables["time_whole"][:], dtype=float)

# variable order: T first, phi second
T_all   = np.array(ds.variables["vals_nod_var1"][:], dtype=float)
phi_all = np.array(ds.variables["vals_nod_var2"][:], dtype=float)

# build grid
x_unique = np.unique(np.round(x, 10))
y_unique = np.unique(np.round(y, 10))

nt = T_all.shape[0]
nx = len(x_unique)
ny = len(y_unique)

def to_grid(vals):
    grid = np.zeros((nt, nx, ny))
    for i in range(len(x)):
        ix = np.argmin(np.abs(x_unique - x[i]))
        iy = np.argmin(np.abs(y_unique - y[i]))
        grid[:, ix, iy] = vals[:, i]
    return grid

T_grid   = to_grid(T_all)
phi_grid = to_grid(phi_all)

# -------------------------
# Split domains
# -------------------------
interface_x = 0.4
fuel_mask   = x_unique <= interface_x + 1e-12
cool_mask   = x_unique >= interface_x - 1e-12

# fuel/coolant data
phi_fuel = phi_grid[:, fuel_mask, :]
T_fuel   = T_grid[:, fuel_mask, :]
T_cool   = T_grid[:, cool_mask, :]

# -------------------------
# Interface slice
# -------------------------
ix_if = np.argmin(np.abs(x_unique - interface_x))
T_interface = T_grid[:, ix_if, :]

# -------------------------
# Save to NPZ
# -------------------------
np.savez(
    "sqp_dataset.npz",
    
    # grids
    t=t,
    x=x_unique,
    y=y_unique,

    # full fields
    phi=phi_grid,
    T=T_grid,

    # split fields (for SQP)
    phi_fuel=phi_fuel,
    Ts=T_fuel,        # Ts = T in fuel
    Tf=T_cool,        # Tf = T in coolant

    # interface
    T_interface=T_interface
)

print("Saved sqp_dataset.npz")
print("phi shape:", phi_grid.shape)
print("T shape:", T_grid.shape)
```
